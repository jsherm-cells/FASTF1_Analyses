```python
import fastf1
import fastf1.plotting
import plotly.express as px
import pandas as pd
import streamlit as st
```

```python
q25 = fastf1.get_session(2025, 'Las Vegas', 'Q')
q25.load()
```

```python
fastf1.plotting.setup_mpl(color_scheme='fastf1')
driver_colors = fastf1.plotting.get_driver_color_mapping(session=q25)
```

```python
q1, q2, q3 = q25.laps.split_qualifying_sessions()
q3
```

```python
q3_nor_25 = q3.pick_drivers('NOR').pick_fastest()
q3_ver_25 = q3.pick_drivers('VER').pick_fastest()
q3_sai_25 = q3.pick_drivers('SAI').pick_fastest()
q3_rus_25 = q3.pick_drivers('RUS').pick_fastest()
q3_pia_25 = q3.pick_drivers('PIA').pick_fastest()
q3_law_25 = q3.pick_drivers('LAW').pick_fastest()
q3_alo_25 = q3.pick_drivers('ALO').pick_fastest()
q3_had_25 = q3.pick_drivers('HAD').pick_fastest()
q3_lec_25 = q3.pick_drivers('LEC').pick_fastest()
q3_gas_25 = q3.pick_drivers('GAS').pick_fastest()
```

```python
q3_nor_25 = q3_nor_25.get_car_data().add_distance()
q3_ver_25 = q3_ver_25.get_car_data().add_distance()
q3_sai_25 = q3_sai_25.get_car_data().add_distance()
q3_rus_25 = q3_rus_25.get_car_data().add_distance()
q3_pia_25 = q3_pia_25.get_car_data().add_distance()
q3_law_25 = q3_law_25.get_car_data().add_distance()
q3_alo_25 = q3_alo_25.get_car_data().add_distance()
q3_had_25 = q3_had_25.get_car_data().add_distance()
q3_lec_25 = q3_lec_25.get_car_data().add_distance()
q3_gas_25 = q3_gas_25.get_car_data().add_distance()
```

```python
q3_nor_25 
```

```python
maskver = (q3_ver_25['Distance'] >= 3200) & (q3_ver_25['Distance'] <= 5300)
q3_ver_25 = q3_ver_25.loc[maskver, ['Throttle', 'Brake', 'Distance', 'Speed', 'nGear', 'RPM', 'DRS', 'Time']]
q3_ver_25['TimeSeconds'] = q3_ver_25['Time'].dt.total_seconds()
```

```python
masksai = (q3_sai_25['Distance'] >= 3200) & (q3_sai_25['Distance'] <= 5300)
q3_sai_25 = q3_sai_25.loc[masksai, ['Throttle', 'Brake', 'Distance', 'Speed', 'nGear', 'RPM', 'DRS', 'Time']]
q3_sai_25['TimeSeconds'] = q3_sai_25['Time'].dt.total_seconds()
```

```python
maskrus = (q3_rus_25['Distance'] >= 3200) & (q3_rus_25['Distance'] <= 5300)
q3_rus_25 = q3_rus_25.loc[maskrus, ['Throttle', 'Brake', 'Distance', 'Speed', 'nGear', 'RPM', 'DRS', 'Time']]
q3_rus_25['TimeSeconds'] = q3_rus_25['Time'].dt.total_seconds()
maskpia = (q3_pia_25['Distance'] >= 3200) & (q3_pia_25['Distance'] <= 5300)
q3_pia_25 = q3_pia_25.loc[maskpia, ['Throttle', 'Brake', 'Distance', 'Speed', 'nGear', 'RPM', 'DRS', 'Time']]
q3_pia_25['TimeSeconds'] = q3_pia_25['Time'].dt.total_seconds()
masklaw = (q3_law_25['Distance'] >= 3200) & (q3_law_25['Distance'] <= 5300)
q3_law_25 = q3_law_25.loc[masklaw, ['Throttle', 'Brake', 'Distance', 'Speed', 'nGear', 'RPM', 'DRS', 'Time']]
q3_law_25['TimeSeconds'] = q3_law_25['Time'].dt.total_seconds()
maskalo = (q3_alo_25['Distance'] >= 3200) & (q3_alo_25['Distance'] <= 5300)
q3_alo_25 = q3_alo_25.loc[maskalo, ['Throttle', 'Brake', 'Distance', 'Speed', 'nGear', 'RPM', 'DRS', 'Time']]
q3_alo_25['TimeSeconds'] = q3_alo_25['Time'].dt.total_seconds()
maskhad = (q3_had_25['Distance'] >= 3200) & (q3_had_25['Distance'] <= 5300)
q3_had_25 = q3_had_25.loc[maskhad, ['Throttle', 'Brake', 'Distance', 'Speed', 'nGear', 'RPM', 'DRS', 'Time']]
q3_had_25['TimeSeconds'] = q3_had_25['Time'].dt.total_seconds()
masklec = (q3_lec_25['Distance'] >= 3200) & (q3_lec_25['Distance'] <= 5300)
q3_lec_25 = q3_lec_25.loc[masklec, ['Throttle', 'Brake', 'Distance', 'Speed', 'nGear', 'RPM', 'DRS', 'Time']]
q3_lec_25['TimeSeconds'] = q3_lec_25['Time'].dt.total_seconds()
maskgas = (q3_gas_25['Distance'] >= 3200) & (q3_gas_25['Distance'] <= 5300)
q3_gas_25 = q3_gas_25.loc[maskgas, ['Throttle', 'Brake', 'Distance', 'Speed', 'nGear', 'RPM', 'DRS', 'Time']]
q3_gas_25['TimeSeconds'] = q3_gas_25['Time'].dt.total_seconds()
```

```python
q25_tele = pd.concat([
    q3_nor_25.assign(Driver='NOR'),
    q3_ver_25.assign(Driver='VER'),
    q3_sai_25.assign(Driver='SAI'),
    q3_rus_25.assign(Driver='RUS'),
    q3_pia_25.assign(Driver='PIA'),
    q3_law_25.assign(Driver='LAW'),
    q3_alo_25.assign(Driver='ALO'),
    q3_had_25.assign(Driver='HAD'),
    q3_lec_25.assign(Driver='LEC'),
    q3_gas_25.assign(Driver='GAS') 
], ignore_index=True)
```

```python
q25_tele['Year'] = 2025
```

```python
q25_tele
```

```python
q25_tele['Speed_ms'] = q25_tele['Speed'] / 3.6
```

```python
deltav = q25_tele.groupby('Driver')['Speed_ms'].diff().dropna()
deltat = q25_tele.groupby('Driver')['TimeSeconds'].diff().dropna()
```

```python
a_g = deltav / deltat / 9.81
```

```python
q25_tele['Longitudinal_Acceleration_g'] = a_g
```

```python
q25_tele = q25_tele.dropna()
q25_tele
```

```python
scat25 = px.scatter(
    q25_tele,
    x='Speed_ms',
    y='Longitudinal_Acceleration_g',
    color='Driver',
    color_discrete_map=driver_colors,
    template='plotly_dark',
    title='Longitudinal Acceleration vs Speed (m/s) - 2025 Las Vegas GP',
    labels={'Speed_ms': 'Speed (m/s)', 'Longitudinal_Acceleration_g': 'Longitudinal Acceleration (g)'},
    trendline='lowess',
    trendline_scope='overall',
)
scat25.update_traces(marker=dict(size=5, opacity=0.7), selector=dict(mode='markers'))
scat25.show()
```

```python
speed25 = px.line(
    q25_tele,
    x='Distance',
    y='Speed',
    color='Driver',
    color_discrete_map=driver_colors,
    template='plotly_dark',
    title='Speed vs Distance (m) - 2025 Las Vegas GP',
    labels={'Distance': 'Distance (m)', 'Speed_ms': 'Speed (m/s)'},
)
speed25.show()
```
